```python
##W7A1: Tìm kiếm nhị phân (Binary Search)
def binary_search(arr, left, right, target):
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1
```

```python
##W7A2: Đếm số lần xuất hiện
def count_occurrences(arr, x):
    return arr.count(x)
```

```python
##W7A3: Sắp xếp (Bubble Sort & Selection Sort)
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr
```

```python
##W7A4: Phần tử xuất hiện nhiều nhất
def most_frequent(arr):
    counts = {}
    first_occurrence = {}
    max_count = 0
    result = arr[0]

    for i, val in enumerate(arr):
        counts[val] = counts.get(val, 0) + 1
        if val not in first_occurrence:
            first_occurrence[val] = i
        
        if counts[val] > max_count:
            max_count = counts[val]
            result = val
        elif counts[val] == max_count:
            # Nếu tần suất bằng nhau, chọn cái xuất hiện trước
            if first_occurrence[val] < first_occurrence[result]:
                result = val
    return f"{result} xuat hien nhieu nhat, som nhat, {max_count} lan"
```

```python
##W7A5: Đếm cặp (a, b) sao cho a + b = X
def count_pairs(arr, X):
    count = 0
    seen = {}
    for num in arr:
        target = X - num
        if target in seen:
            count += seen[target]
        seen[num] = seen.get(num, 0) + 1
    return count
```

```python
##W7A6: Làm phẳng danh sách (Flatten List)
def flatten(nested_list):
    flat_list = []
    for item in nested_list:
        if isinstance(item, list):
            flat_list.extend(flatten(item))
        else:
            flat_list.append(item)
    return flat_list
```

```python
##W7A7: Dãy con tăng nghiêm ngặt dài nhất (LIS)
def length_of_LIS(nums):
    if not nums: return 0
    dp = [1] * len(nums)
    for i in range(len(nums)):
        for j in range(i):
            if nums[i] > nums[j]:
                dp[i] = max(dp[i], dp[j] + 1)
    return max(dp)
```

```python
##W7A8: Tìm khoảng thời gian nhỏ nhất chứa Query
import bisect

def min_interval(intervals, queries):
    # Sắp xếp khoảng theo độ dài (size)
    intervals.sort(key=lambda x: x[1] - x[0])
    results = []
    for q in queries:
        found = -1
        for start, end in intervals:
            if start <= q <= end:
                found = end - start + 1
                break
        results.append(found)
    return results
```

```python
##W7A9: Đếm số nhỏ hơn bên phải
import bisect

def count_smaller(nums):
    sorted_list = []
    res = []
    for x in reversed(nums):
        idx = bisect.bisect_left(sorted_list, x)
        res.append(idx)
        bisect.insort(sorted_list, x)
    return res[::-1]
```

```python
##W7A10: Tổng mảng con lớn nhất modulo m
import bisect

def max_subarray_mod(nums, m):
    prefix = 0
    max_val = 0
    sorted_prefixes = [0]
    for x in nums:
        prefix = (prefix + x) % m
        # Tìm giá trị nhỏ nhất lớn hơn prefix hiện tại trong sorted_prefixes
        idx = bisect.bisect_right(sorted_prefixes, prefix)
        if idx < len(sorted_prefixes):
            max_val = max(max_val, (prefix - sorted_prefixes[idx] + m) % m)
        max_val = max(max_val, prefix)
        bisect.insort(sorted_prefixes, prefix)
    return max_val
```

```python
##W7A12: Sắp xếp thứ tự Ai là triệu phú
def sort_quiz_options(options):
    # options là list 4 chuỗi tương ứng A, B, C, D
    indexed = []
    for i in range(len(options)):
        indexed.append((chr(ord('A') + i), options[i]))
    
    # Sắp xếp theo nội dung chuỗi (item[1])
    indexed.sort(key=lambda x: x[1])
    
    # Trả về chuỗi các ký tự nhãn
    return "".join([item[0] for item in indexed])
```

```python
##W7A13: Dãy hài hòa dài nhất
def longest_harmonious_subsequence(nums):
    counts = {}
    for x in nums:
        counts[x] = counts.get(x, 0) + 1
    
    max_len = 0
    for x in counts:
        if x + 1 in counts:
            max_len = max(max_len, counts[x] + counts[x+1])
    return max_len
```

```python
##W7A14: Tìm vị trí của số 7
def find_sevens(nums):
    indices = [str(i) for i, x in enumerate(nums) if x == 7]
    if not indices:
        return "Not found"
    # Sắp xếp giảm dần theo yêu cầu đề bài
    indices.sort(key=int, reverse=True)
    return " ".join(indices)
```

```python
##W7A15: Bàn tròn Nemo
def get_nemo_neighbors(guests, nemo="Nemo"):
    n = len(guests)
    try:
        idx = guests.index(nemo)
        # Sử dụng modulo để xử lý trường hợp Nemo ở đầu hoặc cuối danh sách
        neighbor_left = guests[(idx - 1) % n]
        neighbor_right = guests[(idx + 1) % n]
        
        # Sắp xếp tên theo thứ tự từ điển để khớp với định dạng Output mẫu
        res = sorted([neighbor_left, neighbor_right])
        return f"{res[0]} and {res[1]}"
    except ValueError:
        return "Nemo not found"
```

```python
##W7A16: Biến đổi thành "Mảng đẹp"
from collections import Counter

def make_beautiful_array(arr):
    counts = Counter(arr)
    deletions = 0
    for x, f in counts.items():
        if x == 0: # Số 0 không thể xuất hiện 0 lần nếu nó đã có mặt
            deletions += f
        elif f >= x:
            deletions += (f - x)
        else:
            deletions += f
    return deletions
```

```python
##W7A17: Thuật toán Boyer–Moore (Majority Element)
##Tìm phần tử chiếm hơn $n/2$ danh sách với $O(n)$ thời gian và $O(1)$ bộ nhớ
def boyer_moore(nums):
    candidate = None
    count = 0
    # Giai đoạn 1: Tìm ứng viên
    for num in nums:
        if count == 0:
            candidate = num
        count += (1 if num == candidate else -1)
    
    # Giai đoạn 2: Kiểm chứng (vì đề bài nói "nếu có")
    if nums.count(candidate) > len(nums) // 2:
        print(candidate)
```

```python
##W7A18: Sắp xếp có điều kiện
##Sử dụng một key tùy chỉnh cho hàm sort()
def conditional_sort(arr):
    # Key: (is_odd, value_for_sorting)
    # Số chẵn: is_odd=False (0), sắp xếp tăng dần
    # Số lẻ: is_odd=True (1), sắp xếp giảm dần (-x)
    arr.sort(key=lambda x: (x % 2 != 0, x if x % 2 == 0 else -x))
    return arr
```

```python
##W7A19: Trộn 2 mảng đã sắp xếp (Two Pointers)
##Không dùng hàm sort, ta dùng hai con trỏ để so sánh từng phần tử.
def merge_sorted_arrays(arr1, arr2):
    # Xác định mảng tăng hay giảm bằng cách so sánh phần tử đầu và cuối
    is_asc = arr1[0] <= arr1[-1] if len(arr1) > 1 else True
    
    res = []
    i, j = 0, 0
    while i < len(arr1) and j < len(arr2):
        condition = (arr1[i] < arr2[j]) if is_asc else (arr1[i] > arr2[j])
        if condition:
            res.append(arr1[i])
            i += 1
        else:
            res.append(arr2[j])
            j += 1
    res.extend(arr1[i:] or arr2[j:])
    return res
```

```python
##W7A21: Tính ngày thứ bao nhiêu của năm 2020
##Năm 2020 là năm nhuận (tháng 2 có 29 ngày).
def day_of_year_2020(day, month):
    days_in_month = [0, 31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    # Kiểm tra tính hợp lệ
    if 1 <= month <= 12 and 1 <= day <= days_in_month[month]:
        return sum(days_in_month[:month]) + day
    return "Ngày không hợp lệ"
```

```python
##W7A22: Tìm tên hay gặp nhất
def most_common_name(names):
    counts = Counter(names)
    # Sắp xếp theo: 1. Tần suất giảm dần (-x[1]), 2. Tên theo alphabet (x[0])
    sorted_names = sorted(counts.items(), key=lambda x: (-x[1], x[0]))
    return sorted_names[0][0]
```

```python
##W7A23: Đếm số lượng tên đúng định dạng
import re

def count_valid_names(text):
    words = text.split()
    # Regex: Chữ cái hoa đứng đầu, theo sau là 1 hoặc nhiều chữ cái thường
    pattern = r'^[A-Z][a-z]+$'
    count = sum(1 for w in words if re.match(pattern, w))
    return count
```

```python
##W7A24: Xóa từ trùng lặp liền kề
def remove_adjacent_duplicates(s):
    words = s.split()
    if not words: return ""
    res = [words[0]]
    for i in range(1, len(words)):
        if words[i] != words[i-1]:
            res.append(words[i])
    return " ".join(res)
```

```python
##W7A25: Số có k chữ số lớn nhất (giữ nguyên thứ tự)
def largest_k_digit_substring(n_str, k):
    max_val = ""
    for i in range(len(n_str) - k + 1):
        current = n_str[i:i+k]
        if current > max_val: # So sánh chuỗi số
            max_val = current
    return max_val
```

```python
##W7A26: Trích xuất số và tính trung bình
import re

def extract_and_stats(text):
    nums = [int(n) for n in re.findall(r'\d+', text)]
    if not nums: return 0, 0
    total = sum(nums)
    avg = total / len(nums)
    return total, avg
```

```python
##W7A27: Kiểm tra dãy ngoặc hợp lệ
def is_valid_parentheses(s):
    stack = []
    mapping = {")": "(", "}": "{", "]": "["}
    for char in s:
        if char in mapping:
            top = stack.pop() if stack else '#'
            if mapping[char] != top:
                return False
        else:
            stack.append(char)
    return not stack
```

```python
##W7A28: Dãy hợp số dài nhất không quá N
def longest_composite_sequence(n):
    def is_prime(k):
        if k < 2: return False
        for i in range(2, int(k**0.5) + 1):
            if k % i == 0: return False
        return True

    composites = [False if is_prime(i) or i < 2 else True for i in range(n + 1)]
    
    max_len = 0
    curr_len = 0
    start_idx = -1
    best_start = -1
    
    for i in range(1, n + 1):
        if composites[i]:
            if curr_len == 0: start_idx = i
            curr_len += 1
            if curr_len > max_len:
                max_len = curr_len
                best_start = start_idx
        else:
            curr_len = 0
            
    return list(range(best_start, best_start + max_len)) if best_start != -1 else []
```

```python
##W7A29: Dãy con tăng dần LIÊN TỤC dài nhất
def longest_continuous_increasing(arr):
    if not arr: return []
    max_len = 1
    curr_len = 1
    best_end = 0
    
    for i in range(1, len(arr)):
        if arr[i] > arr[i-1]:
            curr_len += 1
            if curr_len > max_len:
                max_len = curr_len
                best_end = i
        else:
            curr_len = 1
            
    return arr[best_end - max_len + 1 : best_end + 1]
```

```python
##W7A30: Dãy con tăng dần KHÔNG liên tục dài nhất (LIS)
import bisect

def longest_increasing_subsequence(arr):
    if not arr: return 0
    tails = []
    for x in arr:
        idx = bisect.bisect_left(tails, x)
        if idx < len(tails):
            tails[idx] = x
        else:
            tails.append(x)
    return len(tails)
```
