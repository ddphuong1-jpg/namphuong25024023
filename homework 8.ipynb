```python
# =============================================================================
# ĐỀ BÀI W8A16:
# Yêu cầu:
# Xây dựng lớp Shape với phương thức area().
# Xây dựng lớp Rectangle(w, h) và Circle(r) kế thừa từ Shape, ghi đè area().
# Lấy pi = 3.14.
# Input:
# - Dòng 1: loại hình (RECTANGLE hoặc CIRCLE)
# - Dòng 2: tham số tương ứng.
# Output: In ra diện tích với 2 chữ số thập phân.
# =============================================================================

class Shape:
    def area(self):
        return 0.0

class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h
        
    def area(self):
        return self.w * self.h

class Circle(Shape):
    def __init__(self, r):
        self.r = r
        
    def area(self):
        return 3.14 * self.r * self.r

if __name__ == "__main__":
    shape_type = input().strip()
    params = list(map(float, input().split()))
    
    if shape_type == "RECTANGLE":
        rect = Rectangle(params[0], params[1])
        print(f"{rect.area():.2f}")
    elif shape_type == "CIRCLE":
        circle = Circle(params[0])
        print(f"{circle.area():.2f}")
```

```python
# =============================================================================
# ĐỀ BÀI W8A17:
# Yêu cầu:
# Xây dựng lớp Vehicle(brand, speed).
# Xây dựng lớp Car(brand, speed, seats) kế thừa từ Vehicle.
# Viết phương thức info() cho mỗi lớp:
# - Vehicle: brand-speed
# - Car: brand-speed-seats
# Input:
# - Dòng 1: loại đối tượng (Vehicle hoặc Car)
# - Dòng 2: dữ liệu tương ứng.
# Output: In ra chuỗi info().
# =============================================================================

class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed
        
    def info(self):
        return f"{self.brand}-{self.speed}"

class Car(Vehicle):
    def __init__(self, brand, speed, seats):
        super().__init__(brand, speed)
        self.seats = seats
        
    def info(self):
        return f"{self.brand}-{self.speed}-{self.seats}"

if __name__ == "__main__":
    obj_type = input().strip()
    data = input().split()
    
    if obj_type == "Vehicle":
        brand = data[0]
        speed = data[1]
        v = Vehicle(brand, speed)
        print(v.info())
    elif obj_type == "Car":
        brand = data[0]
        speed = data[1]
        seats = data[2]
        c = Car(brand, speed, seats)
        print(c.info())
```

```python
# =============================================================================
# ĐỀ BÀI W8A18:
# Yêu cầu:
# Xây dựng lớp Polygon với phương thức perimeter().
# Xây dựng lớp Triangle(a, b, c) kế thừa từ Polygon và ghi đè perimeter() trả về a+b+c.
# Input: Một dòng chứa 3 số nguyên a b c.
# Output: In ra chu vi tam giác.
# =============================================================================

class Polygon:
    def perimeter(self):
        return 0

class Triangle(Polygon):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
        
    def perimeter(self):
        return self.a + self.b + self.c

if __name__ == "__main__":
    a, b, c = map(int, input().split())
    tri = Triangle(a, b, c)
    print(tri.perimeter())
```

```python
# =============================================================================
# ĐỀ BÀI W8A19:
# Yêu cầu:
# Xây dựng lớp Person(name, age) và lớp Student(name, age, school) kế thừa từ Person.
# Phương thức describe():
# - Person: name age
# - Student: name age school
# Input:
# - Dòng 1: Person hoặc Student
# - Dòng 2: dữ liệu tương ứng.
# Output: In ra kết quả describe().
# =============================================================================

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def describe(self):
        return f"{self.name} {self.age}"

class Student(Person):
    def __init__(self, name, age, school):
        super().__init__(name, age)
        self.school = school
        
    def describe(self):
        return f"{self.name} {self.age} {self.school}"

if __name__ == "__main__":
    obj_type = input().strip()
    data = input().split()
    
    if obj_type == "Person":
        name = data[0]
        age = data[1]
        p = Person(name, age)
        print(p.describe())
    elif obj_type == "Student":
        name = data[0]
        age = data[1]
        school = data[2]
        s = Student(name, age, school)
        print(s.describe())
```

```python
# =============================================================================
# ĐỀ BÀI W8A20:
# Yêu cầu:
# Xây dựng lớp Account(balance) với hai phương thức deposit(x) và get_balance().
# Xây dựng lớp SavingAccount(balance, rate) kế thừa từ Account, thêm phương thức
# add_interest() để cộng balance*rate vào số dư.
# Input: balance rate
# Output: Số dư sau khi cộng lãi, in 2 chữ số thập phân.
# =============================================================================

class Account:
    def __init__(self, balance):
        self._balance = float(balance)
        
    def deposit(self, x):
        self._balance += float(x)
        
    def get_balance(self):
        return self._balance

class SavingAccount(Account):
    def __init__(self, balance, rate):
        super().__init__(balance)
        self.rate = float(rate)
        
    def add_interest(self):
        interest = self._balance * self.rate
        self.deposit(interest)

if __name__ == "__main__":
    balance, rate = map(float, input().split())
    sa = SavingAccount(balance, rate)
    sa.add_interest()
    print(f"{sa.get_balance():.2f}")
```

```python
# =============================================================================
# ĐỀ BÀI W8A21:
# Yêu cầu:
# Xây dựng lớp Printer với phương thức print_value(x).
# Xây dựng lớp BinaryPrinter kế thừa từ Printer, ghi đè print_value(x) để in biểu diễn
# nhị phân của số nguyên dương x.
# Input: Một số nguyên dương n.
# Output: In ra dạng nhị phân của n.
# =============================================================================

class Printer:
    def print_value(self, x):
        print(x)

class BinaryPrinter(Printer):
    def print_value(self, x):
        # bin(x) trả về chuỗi '0b...', dùng [2:] để bỏ phần '0b'
        print(bin(x)[2:])

if __name__ == "__main__":
    n = int(input().strip())
    bp = BinaryPrinter()
    bp.print_value(n)
```

```python
# =============================================================================
# ĐỀ BÀI W8A22:
# Yêu cầu:
# Xây dựng lớp Matrix2x2(a, b, c, d) biểu diễn ma trận:
# [a b]
# [c d]
# Viết phương thức trace() trả về a + d và determinant() trả về ad - bc.
# Input: Một dòng chứa 4 số nguyên a b c d.
# Output: In ra trace và determinant trên cùng một dòng.
# =============================================================================

class Matrix2x2:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d
        
    def trace(self):
        return self.a + self.d
        
    def determinant(self):
        return (self.a * self.d) - (self.b * self.c)

if __name__ == "__main__":
    a, b, c, d = map(int, input().split())
    matrix = Matrix2x2(a, b, c, d)
    print(f"{matrix.trace()} {matrix.determinant()}")
```

```python
# =============================================================================
# ĐỀ BÀI W8A23:
# Yêu cầu:
# Xây dựng lớp Polynomial biểu diễn đa thức bậc nhất ax + b.
# Viết phương thức evaluate(x) để tính giá trị đa thức tại x.
# Input: a b x
# Output: Giá trị của ax + b.
# =============================================================================

class Polynomial:
    def __init__(self, a, b):
        self.a = a
        self.b = b
        
    def evaluate(self, x):
        return self.a * x + self.b

if __name__ == "__main__":
    a, b, x = map(float, input().split())
    poly = Polynomial(a, b)
    res = poly.evaluate(x)
    # Rút gọn định dạng nếu là số nguyên tròn
    print(int(res) if res.is_integer() else res)
```

```python
# =============================================================================
# ĐỀ BÀI W8A24:
# Yêu cầu:
# Xây dựng lớp Polynomial2 biểu diễn đa thức bậc hai ax^2 + bx + c.
# Viết phương thức evaluate(x) để tính giá trị đa thức tại x.
# Input: a b c x
# Output: Giá trị của đa thức.
# =============================================================================

class Polynomial2:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
        
    def evaluate(self, x):
        return self.a * (x ** 2) + self.b * x + self.c

if __name__ == "__main__":
    a, b, c, x = map(float, input().split())
    poly2 = Polynomial2(a, b, c)
    res = poly2.evaluate(x)
    print(int(res) if res.is_integer() else res)
```

```python
# =============================================================================
# ĐỀ BÀI W8A25:
# Yêu cầu:
# Xây dựng lớp PolynomialSum cho hai đa thức bậc nhất:
# P(x) = ax + b, Q(x) = cx + d.
# Viết phương thức sum_str() trả về chuỗi biểu diễn đa thức tổng dưới dạng kx + m.
# Input: a b c d
# Output: Chuỗi của đa thức P(x)+Q(x).
# =============================================================================

class PolynomialSum:
    def __init__(self, a, b, c, d):
        self.k = a + c
        self.m = b + d
        
    def sum_str(self):
        if self.m > 0:
            return f"{self.k}x + {self.m}"
        elif self.m < 0:
            return f"{self.k}x - {abs(self.m)}"
        else:
            return f"{self.k}x"

if __name__ == "__main__":
    a, b, c, d = map(int, input().split())
    poly_sum = PolynomialSum(a, b, c, d)
    print(poly_sum.sum_str())
```

```python
# =============================================================================
# ĐỀ BÀI W8A27:
# Yêu cầu:
# Xây dựng lớp Product(name, price).
# Xây dựng lớp DiscountProduct(name, price, percent) kế thừa từ Product,
# ghi đè phương thức final_price() để trả về giá sau giảm.
# Công thức: price * (100 - percent) / 100
# Input: name price percent
# Output: In ra giá cuối cùng với 2 chữ số thập phân.
# =============================================================================

class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = float(price)
        
    def final_price(self):
        return self.price

class DiscountProduct(Product):
    def __init__(self, name, price, percent):
        super().__init__(name, price)
        self.percent = float(percent)
        
    def final_price(self):
        return self.price * (100 - self.percent) / 100

if __name__ == "__main__":
    data = input().split()
    name = data[0]
    price = float(data[1])
    percent = float(data[2])
    
    dp = DiscountProduct(name, price, percent)
    print(f"{dp.final_price():.2f}")
```

```python
# =============================================================================
# ĐỀ BÀI W8A29:
# Yêu cầu:
# Xây dựng lớp Time(h, m, s) với phương thức to_seconds() đổi thời gian sang tổng số giây.
# Input: h m s
# Output: Tổng số giây.
# =============================================================================

class Time:
    def __init__(self, h, m, s):
        self.h = h
        self.m = m
        self.s = s
        
    def to_seconds(self):
        return self.h * 3600 + self.m * 60 + self.s

if __name__ == "__main__":
    h, m, s = map(int, input().split())
    t = Time(h, m, s)
    print(t.to_seconds())
```

```python
# =============================================================================
# ĐỀ BÀI W8A28:
# Yêu cầu:
# Xây dựng lớp Counter có thuộc tính count ban đầu bằng 0 và các phương thức inc(), dec(), value().
# Sau đó thực hiện dãy thao tác trên đối tượng.
# Input:
# - Dòng 1: n
# - Dòng 2: n thao tác, mỗi thao tác là INC hoặc DEC, cách nhau bởi dấu cách.
# Output: Giá trị cuối cùng của count.
# =============================================================================

class Counter:
    def __init__(self):
        self.count = 0
        
    def inc(self):
        self.count += 1
        
    def dec(self):
        self.count -= 1
        
    def value(self):
        return self.count

if __name__ == "__main__":
    n = int(input().strip())
    operations = input().split()
    
    counter = Counter()
    for op in operations:
        if op == "INC":
            counter.inc()
        elif op == "DEC":
            counter.dec()
            
    print(counter.value())
```

```python
# =============================================================================
# ĐỀ BÀI W8A30:
# Yêu cầu:
# Xây dựng lớp Stack sử dụng danh sách để cài đặt ngăn xếp với các phương thức
# push(x), pop(), top(). Dữ liệu đảm bảo mọi lệnh pop và top đều hợp lệ.
# Input:
# - Dòng 1: n
# - n dòng tiếp theo là một trong ba lệnh: PUSH x, POP, TOP.
# Output: Mỗi lệnh TOP in ra giá trị ở đỉnh ngăn xếp trên một dòng.
# =============================================================================

class Stack:
    def __init__(self):
        self.items = []
        
    def push(self, x):
        self.items.append(x)
        
    def pop(self):
        if not self.is_empty():
            return self.items.pop()
            
    def top(self):
        if not self.is_empty():
            return self.items[-1]
            
    def is_empty(self):
        return len(self.items) == 0

if __name__ == "__main__":
    n = int(input().strip())
    stack = Stack()
    
    for _ in range(n):
        command = input().split()
        if command[0] == "PUSH":
            stack.push(int(command[1]))
        elif command[0] == "POP":
            stack.pop()
        elif command[0] == "TOP":
            print(stack.top())
```
