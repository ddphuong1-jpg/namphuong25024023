```python
# ĐỀ BÀI VẮN TẮT:
# - Tạo lớp Shape có phương thức area().
# - Lớp Rectangle(w, h) và Circle(r) kế thừa từ Shape và ghi đè area(). Lấy pi = 3.14.
# - Input: Dòng 1 là loại hình (RECTANGLE/CIRCLE), Dòng 2 là tham số tương ứng.
# - Output: Diện tích làm tròn 2 chữ số thập phân.

import sys

class Shape:
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h

class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14 * self.r * self.r

def solve_w8a16():
    lines = sys.stdin.read().splitlines()
    if not lines: return
    shape_type = lines[0].strip()
    params = list(map(float, lines[1].split()))
    
    if shape_type == "RECTANGLE":
        obj = Rectangle(params[0], params[1])
    else:
        obj = Circle(params[0])
        
    print(f"{obj.area():.2f}")

solve_w8a16()
```

```python
# ĐỀ BÀI VẮN TẮT:
# - Tạo lớp Polygon có phương thức perimeter().
# - Lớp Triangle(a, b, c) kế thừa từ Polygon, ghi đè perimeter() trả về a + b + c.
# - Input: Một dòng chứa 3 số nguyên a, b, c.
# - Output: Chu vi tam giác.

import sys

class Polygon:
    def perimeter(self):
        pass

class Triangle(Polygon):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    def perimeter(self):
        return self.a + self.b + self.c

def solve_w8a18():
    line = sys.stdin.read().strip()
    if not line: return
    a, b, c = map(int, line.split())
    t = Triangle(a, b, c)
    print(t.perimeter())

solve_w8a18()
```

```python
# ĐỀ BÀI VẮN TẮT:
# - Tạo lớp Account(balance) có deposit(x) và get_balance().
# - Lớp SavingAccount(balance, rate) kế thừa từ Account, thêm add_interest() để cộng balance * rate vào số dư.
# - Input: Một dòng chứa balance và rate.
# - Output: Số dư sau khi cộng lãi, in với 2 chữ số thập phân.

import sys

class Account:
    def __init__(self, balance):
        self.balance = balance
    def deposit(self, x):
        self.balance += x
    def get_balance(self):
        return self.balance

class SavingAccount(Account):
    def __init__(self, balance, rate):
        super().__init__(balance)
        self.rate = rate
    def add_interest(self):
        self.balance += self.balance * self.rate

def solve_w8a20():
    line = sys.stdin.read().strip()
    if not line: return
    balance, rate = map(float, line.split())
    acc = SavingAccount(balance, rate)
    acc.add_interest()
    print(f"{acc.get_balance():.2f}")

solve_w8a20()
```

```python
# ĐỀ BÀI VẮN TẮT:
# - Tạo lớp Printer có phương thức print_value(x).
# - Tạo lớp BinaryPrinter kế thừa Printer, ghi đè print_value(x) để trả về biểu diễn nhị phân của số nguyên dương x.
# - Input: Một số nguyên dương n.
# - Output: Chuỗi nhị phân của n.

import sys

class Printer:
    def print_value(self, x):
        pass

class BinaryPrinter(Printer):
    def print_value(self, x):
        return bin(x)[2:]

def solve_w8a21():
    line = sys.stdin.read().strip()
    if not line: return
    n = int(line)
    bp = BinaryPrinter()
    print(bp.print_value(n))

solve_w8a21()
```

```python
# ĐỀ BÀI VẮN TẮT:
# - Tạo lớp Matrix2x2(a, b, c, d) biểu diễn ma trận 2x2 hàng 1 [a, b], hàng 2 [c, d].
# - Viết phương thức trace() trả về a + d và determinant() trả về ad - bc.
# - Input: Một dòng chứa 4 số nguyên a, b, c, d.
# - Output: Giá trị trace và determinant trên cùng một dòng.

import sys

class Matrix2x2:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d
    def trace(self):
        return self.a + self.d
    def determinant(self):
        return self.a * self.d - self.b * self.c

def solve_w8a22():
    line = sys.stdin.read().strip()
    if not line: return
    a, b, c, d = map(int, line.split())
    m = Matrix2x2(a, b, c, d)
    print(f"{m.trace()} {m.determinant()}")

solve_w8a22()
```

```python
# ĐỀ BÀI VẮN TẮT:
# - Tạo lớp PolynomialSum cho 2 đa thức bậc nhất P(x) = ax + b, Q(x) = cx + d.
# - Viết phương thức sum_str() trả về chuỗi biểu diễn tổng dưới dạng "kx + m".
# - Input: 4 số nguyên a, b, c, d.
# - Output: Chuỗi tổng đa thức P(x) + Q(x).

import sys

class PolynomialSum:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d
    def sum_str(self):
        k = self.a + self.c
        m = self.b + self.d
        return f"{k}x + {m}"

def solve_w8a25():
    line = sys.stdin.read().strip()
    if not line: return
    a, b, c, d = map(int, line.split())
    p_sum = PolynomialSum(a, b, c, d)
    print(p_sum.sum_str())

solve_w8a25()
```

```python
# ĐỀ BÀI VẮN TẮT:
# - Tạo lớp Product(name, price) và DiscountProduct(name, price, percent) kế thừa từ Product.
# - Ghi đè/thêm phương thức final_price() trả về giá sau giảm: price * (100 - percent) / 100.
# - Input: Một dòng chứa name, price, percent.
# - Output: Giá cuối cùng sau khi giảm, in với 2 chữ số thập phân.

import sys

class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

class DiscountProduct(Product):
    def __init__(self, name, price, percent):
        super().__init__(name, price)
        self.percent = percent
    def final_price(self):
        return self.price * (100 - self.percent) / 100

def solve_w8a27():
    line = sys.stdin.read().strip()
    if not line: return
    parts = line.split()
    name = parts[0]
    price = float(parts[1])
    percent = float(parts[2])
    
    dp = DiscountProduct(name, price, percent)
    print(f"{dp.final_price():.2f}")

solve_w8a27()
```

```python
# ĐỀ BÀI VẮN TẮT:
# - Tạo lớp Counter có thuộc tính count ban đầu bằng 0 và các phương thức inc(), dec(), value().
# - Thực hiện một dãy thao tác INC hoặc DEC trên đối tượng.
# - Input: Dòng 1 chứa số lượng thao tác n, Dòng 2 chứa n thao tác cách nhau bằng khoảng trắng.
# - Output: Giá trị cuối cùng của count.

import sys

class Counter:
    def __init__(self):
        self.count = 0
    def inc(self):
        self.count += 1
    def dec(self):
        self.count -= 1
    def value(self):
        return self.count

def solve_w8a28():
    lines = sys.stdin.read().splitlines()
    if len(lines) < 2: return
    operations = lines[1].split()
    
    counter = Counter()
    for op in operations:
        if op == "INC":
            counter.inc()
        elif op == "DEC":
            counter.dec()
            
    print(counter.value())

solve_w8a28()
```

```python
# ĐỀ BÀI VẮN TẮT:
# - Tạo lớp Time(h, m, s) với phương thức to_seconds() đổi thời gian sang tổng số giây.
# - Input: 3 số nguyên h m s.
# - Output: Tổng số giây.

import sys

class Time:
    def __init__(self, h, m, s):
        self.h = h
        self.m = m
        self.s = s
    def to_seconds(self):
        return self.h * 3600 + self.m * 60 + self.s

def solve_w8a29():
    line = sys.stdin.read().strip()
    if not line: return
    h, m, s = map(int, line.split())
    t = Time(h, m, s)
    print(t.to_seconds())

solve_w8a29()
```

```python
# ĐỀ BÀI VẮN TẮT:
# - Tạo lớp Stack sử dụng list để cài đặt ngăn xếp với các phương thức push(x), pop(), top().
# - Input: Dòng 1 chứa số lệnh n. n dòng tiếp theo là các lệnh PUSH x, POP, hoặc TOP.
# - Output: Mỗi lệnh TOP in ra giá trị ở đỉnh ngăn xếp trên một dòng độc lập.

import sys

class Stack:
    def __init__(self):
        self.items = []
    def push(self, x):
        self.items.append(x)
    def pop(self):
        if self.items:
            return self.items.pop()
    def top(self):
        if self.items:
            return self.items[-1]

def solve_w8a30():
    input_data = sys.stdin.read().splitlines()
    if not input_data: return
    
    n = int(input_data[0].strip())
    stack = Stack()
    
    for i in range(1, n + 1):
        if i >= len(input_data): break
        command = input_data[i].strip().split()
        if not command: continue
        
        cmd_type = command[0]
        if cmd_type == "PUSH":
            stack.push(int(command[1]))
        elif cmd_type == "POP":
            stack.pop()
        elif cmd_type == "TOP":
            print(stack.top())

solve_w8a30()
```
