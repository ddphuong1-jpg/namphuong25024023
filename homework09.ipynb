```python
#W9A16: In chuỗi nhị phân có n số 0 và m số 1
def generate_binary(n, m, current_string=""):
    # Nếu hết cả 0 và 1 thì in ra cấu hình hiện tại
    if n == 0 and m == 0:
        print(current_string)
        return
    
    # Ưu tiên thêm '0' trước để đảm bảo thứ tự từ điển
    if n > 0:
        generate_binary(n - 1, m, current_string + "0")
        
    # Thêm '1' sau
    if m > 0:
        generate_binary(n, m - 1, current_string + "1")

# Test thử nghiệm với n = 2, m = 1
print("--- Kết quả W9A16 ---")
generate_binary(2, 1)
```

```python
#W9A17: Dãy con tăng nghiêm ngặt dài nhất
def find_lis(arr):
    if not arr:
        return []
        
    n = len(arr)
    # L[i] lưu độ dài dãy con tăng dài nhất kết thúc tại arr[i]
    L = [1] * n
    # trace[i] lưu chỉ số của phần tử đứng trước arr[i] trong dãy con
    trace = [-1] * n
    
    for i in range(1, n):
        for j in range(0, i):
            if arr[i] > arr[j] and L[i] < L[j] + 1:
                L[i] = L[j] + 1
                trace[i] = j
                
    # Tìm vị trí kết thúc của dãy con dài nhất
    max_len = max(L)
    max_idx = L.index(max_len)
    
    # Truy vết ngược lại
    result = []
    while max_idx != -1:
        result.append(arr[max_idx])
        max_idx = trace[max_idx]
        
    return result[::-1]

print("--- Kết quả W9A17 ---")
arr_input = [10, 22, 9, 33, 21, 50, 41, 60, 80]
print("Dãy ban đầu:", arr_input)
print("Dãy con tăng dài nhất:", find_lis(arr_input))
```

```python
#W9A18: Đếm số cặp có tổng bằng k
def count_pairs_with_sum(arr, k):
    freq = {}
    count = 0
    
    for x in arr:
        target = k - x
        # Nếu target đã xuất hiện trước đó, cộng số lần xuất hiện vào biến đếm
        if target in freq:
            count += freq[target]
            
        # Lưu trữ tần suất xuất hiện của x
        freq[x] = freq.get(x, 0) + 1
        
    return count

print("--- Kết quả W9A18 ---")
arr_test = [1, 5, 7, -1, 5]
k_test = 6
print(f"Số cặp có tổng bằng {k_test}:", count_pairs_with_sum(arr_test, k_test))
```

```python
#W9A19: Loại bỏ phần tử trùng lặp giữ nguyên thứ tự
def remove_duplicates(arr):
    seen = set()
    result = []
    for item in arr:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result

print("--- Kết quả W9A19 ---")
duplicate_arr = [3, 1, 2, 3, 2, 4, 1]
print("Mảng gốc:", duplicate_arr)
print("Sau khi loại trùng:", remove_duplicates(duplicate_arr))
```

```python
#W9A20: Chuẩn hóa chuỗi kí tự
def custom_normalize(s):
    # Bước 1: Tìm vị trí thực sự không phải khoảng trắng ở đầu và cuối (Xử lý câu a)
    start = 0
    while start < len(s) and s[start] == ' ':
        start += 1
        
    end = len(s) - 1
    while end >= 0 and s[end] == ' ':
        end -= 1
        
    if start > end:
        return ""
        
    res = []
    is_new_word = True  # Đánh dấu ký tự đầu tiên của từ
    
    # Bước 2 & 3: Duyệt từ start đến end để xóa dấu cách thừa và viết hoa (Xử lý câu b, c)
    for i in range(start, end + 1):
        char = s[i]
        
        if char == ' ':
            # Nếu ký tự hiện tại là khoảng trắng và ký tự trước đó không phải khoảng trắng
            if len(res) > 0 and res[-1] != ' ':
                res.append(' ')
            is_new_word = True
        else:
            # Chuyển ký tự thành mã ASCII tương ứng
            ascii_code = ord(char)
            
            if is_new_word:
                # Nếu là chữ thường (a-z), đổi thành chữ hoa (A-Z)
                if 97 <= ascii_code <= 122:
                    char = chr(ascii_code - 32)
                is_new_word = False
            else:
                # Nếu là chữ hoa (A-Z) nhưng không phải đầu từ, đổi thành chữ thường (a-z)
                if 65 <= ascii_code <= 90:
                    char = chr(ascii_code + 32)
                    
            res.append(char)
            
    return "".join(res)

print("--- Kết quả W9A20 ---")
raw_str = "   vũ   văn   tHảO   "
print(f"Gốc: '{raw_str}'")
print(f"Sau chuẩn hóa: '{custom_normalize(raw_str)}'")
```

```python
#W9A21: Tìm phần tử lớn thứ k bằng Quickselect
def quickselect(arr, left, right, k_small):
    if left == right:
        return arr[left]
        
    # Chọn pivot là phần tử ở giữa
    pivot = arr[(left + right) // 2]
    
    # Phân hoạch mảng
    i, j = left, right
    while i <= j:
        while arr[i] < pivot: i += 1
        while arr[j] > pivot: j -= 1
        if i <= j:
            arr[i], arr[j] = arr[j], arr[i]
            i += 1
            j -= 1
            
    # k_small nằm bên vùng nào thì đệ quy vùng đó
    if k_small <= j:
        return quickselect(arr, left, j, k_small)
    if k_small >= i:
        return quickselect(arr, i, right, k_small)
        
    return arr[k_small]

def find_kth_largest(arr, k):
    n = len(arr)
    # Phần tử lớn thứ k chính là phần tử nhỏ thứ (n - k) trong mảng tăng dần
    return quickselect(arr, 0, n - 1, n - k)

print("--- Kết quả W9A21 ---")
arr_q = [3, 2, 1, 5, 6, 4]
k_largest = 2
print(f"Phần tử lớn thứ {k_largest} là:", find_kth_largest(arr_q.copy(), k_largest))
```

```python
#Kiểm tra hai chuỗi Anagram
def is_anagram(str1, str2):
    # Loại bỏ khoảng trắng và chuyển về chữ thường để so sánh chính xác nếu cần
    str1 = str1.replace(" ", "").lower()
    str2 = str2.replace(" ", "").lower()
    
    if len(str1) != len(str2):
        return False
        
    count = {}
    for char in str1:
        count[char] = count.get(char, 0) + 1
        
    for char in str2:
        if char not in count:
            return False
        count[char] -= 1
        if count[char] < 0:
            return False
            
    return True

print("--- Kết quả W9A22 ---")
s1, s2 = "silent", "listen"
print(f"'{s1}' và '{s2}' có phải anagram?", is_anagram(s1, s2))
```

```python
#W9A23: Tìm số có k chữ số lớn nhất bằng cách xóa bớt chữ số
def max_number_by_removing(s, k):
    rem = len(s) - k  # Số lượng chữ số cần phải xóa bỏ
    stack = []
    
    for digit in s:
        # Trong khi vẫn còn lượt xóa và số hiện tại lớn hơn số ở đỉnh stack
        while rem > 0 and stack and stack[-1] < digit:
            stack.pop()
            rem -= 1
        stack.append(digit)
        
    # Nếu duyệt hết chuỗi mà vẫn chưa xóa đủ số lượt (ví dụ chuỗi giảm dần)
    while rem > 0:
        stack.pop()
        rem -= 1
        
    return "".join(stack)

print("--- Kết quả W9A23 ---")
num_str = "78934340999"
k_digits = 7
print(f"Số lớn nhất gồm {k_digits} chữ số từ {num_str} là:", max_number_by_removing(num_str, k_digits))
```

```python
#W9A24: Nén chuỗi ký tự liên tiếp
def compress_string(s):
    if not s:
        return ""
        
    result = []
    count = 1
    
    for i in range(1, len(s)):
        if s[i] == s[i-1]:
            count += 1
        else:
            result.append(f"{s[i-1]}{count}")
            count = 1
            
    # Thêm ký tự cuối cùng vào danh sách
    result.append(f"{s[-1]}{count}")
    
    return "".join(result)

print("--- Kết quả W9A24 ---")
string_to_compress = "aaaaaafdhhhhhh"
print("Chuỗi gốc:", string_to_compress)
print("Chuỗi sau khi nén:", compress_string(string_to_compress))
```

```python
#W9A25: Kiểm tra mật khẩu mạnh
def check_strong_password(password):
    if len(password) < 8:
        return False
        
    has_lower = False
    has_upper = False
    has_digit = False
    has_special = False
    
    special_chars = {'!', '@', '#', '$'}
    
    for char in password:
        if char.islower():
            has_lower = True
        elif char.isupper():
            has_upper = True
        elif char.isdigit():
            has_digit = True
        elif char in special_chars:
            has_special = True
            
    return has_lower and has_upper and has_digit and has_special

print("--- Kết quả W9A25 ---")
pass1 = "Uet2026!"
pass2 = "12345abc"
print(f"Mật khẩu '{pass1}' có mạnh không?", check_strong_password(pass1))
print(f"Mật khẩu '{pass2}' có mạnh không?", check_strong_password(pass2))
```
